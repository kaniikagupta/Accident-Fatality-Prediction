```python
#Seats
def categorize_seat_revised(position):
    if pd.isna(position):
        return 'Front_Passenger'
    
    position_std = str(position).strip().upper()
    
    high_risk_keywords = [
        'EXTERIOR', 'TRAILING UNIT', 'SLEEPER SECTION', 'APPENDED',
        'ENCLOSED PASSENGER OR CARGO AREA',  
    ]
    if any(keyword in position_std for keyword in high_risk_keywords):
        return 'Non_Standard_Risk'
   
    if position_std == 'FRONT SEAT, LEFT SIDE':
        return 'Driver'
    
    elif 'FRONT SEAT' in position_std:
        return 'Front_Passenger'
    
    elif 'SECOND SEAT' in position_std or 'THIRD SEAT' in position_std or 'FOURTH SEAT' in position_std:
        return 'Rear_Seat'

    return 'Front_Passenger'

vehicle_person['Seat_Category'] = vehicle_person['Seat_Position'].apply(categorize_seat_revised)
vehicle_person = vehicle_person.drop(columns=['Seat_Position'])

vehicle_person['Killed_Seat_Position'] = vehicle_person.apply(
    lambda row: row['Seat_Category'] if row['Is_Killed'] == 1 else 'Not Applicable',
    axis=1
)

def get_fatal_seats_list(series):
    fatal_seats = series[series != 'Not Applicable']
    return fatal_seats.tolist()

total_killed_alive = vehicle_person.groupby('Crash_ID').agg(
    Total_Killed=('Is_Killed', 'sum'),
    Total_Alive=('Is_Alive', 'sum'),
    Fatal_Seats_List=('Killed_Seat_Position', get_fatal_seats_list)
).reset_index()

print(total_killed_alive.head())
```

```python
#Restraint_Use

def categorize_restraint(restraint):
    if pd.isna(restraint):
        return 'Shoulder_Lapbelt'
    
    restraint_std = str(restraint).strip().upper()

    if 'SHOULDER AND LAP BELT USED' in restraint_std:
        return 'Shoulder_Lapbelt'
    elif 'CHILD' in restraint_std or 'BOOSTER SEAT' in restraint_std:
        return 'Child_Safety'
    elif 'SHOULDER BELT ONLY USED' in restraint_std:
        return 'Shoulder_Only'
    elif 'LAP BELT ONLY USED' in restraint_std:
        return 'Lapbelt_Only'
    elif 'NONE USED' in restraint_std:
        return 'None_Used'
    return 'Shoulder_Lapbelt'

vehicle_person['Restraint_Category'] = vehicle_person['Restraint_Use'].apply(categorize_restraint)
vehicle_person = vehicle_person.drop(columns=['Restraint_Use'])

vehicle_person['Killed_Restraint_Status'] = vehicle_person.apply(
    lambda row: row['Restraint_Category'] if row['Is_Killed'] == 1 else 'Not Applicable',
    axis=1
)

restraint_features = vehicle_person.groupby('Crash_ID').agg(
    Killed_Restraint_List=('Killed_Restraint_Status', get_fatal_seats_list)
).reset_index()

total_killed_alive = total_killed_alive.merge(
    restraint_features,
    on='Crash_ID',
    how='left' 
)
```

```python
import numpy as np
def categorize_license(status):
    if pd.isna(status):
        return 'Invalid'
    
    status_std = str(status).strip().upper()

    if status_std == 'VALID':
        return 'Valid'
    elif status_std in ['NOT LICENSED', 'SUSPENDED', 'EXPIRED', 'REVOKED', 'CANCELED OR DENIED']:
        return 'Invalid'
    else:
        return 'Invalid'

vehicle_person['License_Category'] = vehicle_person['License_Status'].apply(categorize_license)
vehicle_person = vehicle_person.drop(columns=['License_Status'])

vehicle_person['Driver_License_Category'] = np.where(
    vehicle_person['Seat_Category'] == 'Driver', 
    vehicle_person['License_Category'], 
    np.nan
)

history_cols = ['Previous_Accident_Flag', 'Previous_Speeding_Flag']

for col in history_cols:
    numeric_values = pd.to_numeric(vehicle_person[col], errors='coerce')
    vehicle_person[col] = numeric_values.fillna(0).astype(int)

driver_subset = vehicle_person[vehicle_person['Seat_Category'] == 'Driver'].copy()

driver_history_features = driver_subset.groupby('Crash_ID').agg(
    Driver_License_Category=('Driver_License_Category', 'first'),
    Driver_Max_Prev_Accident_Count=('Previous_Accident_Flag', 'max'),
    Driver_Max_Prev_Speeding_Count=('Previous_Speeding_Flag', 'max')
).reset_index()

total_killed_alive = total_killed_alive.merge(
    driver_history_features, 
    on='Crash_ID', 
    how='left'
)
```

```python
#Speeding
def categorize_speeding(status):
    if pd.isna(status):
        return 'No'
    
    status_std = str(status).strip().upper()

    if status_std == 'NO':
        return 'No'
    elif 'YES' in status_std or 'RACING' in status_std:
        return 'Yes'
    elif 'TOO FAST FOR CONDITIONS' in status_std:
        return 'Too Fast' 
    else:
        return 'No'

vehicle_person['Speeding_Status_Category'] = vehicle_person['Speed_Related_Flag'].apply(categorize_speeding)
vehicle_person = vehicle_person.drop(columns=['Speed_Related_Flag'])

def is_exceeding_limit(status):
    category = categorize_speeding(status)
    return 1 if category == 'Yes' else 0

def is_too_fast_for_conditions(status):
    category = categorize_speeding(status)
    return 1 if category == 'Too Fast' else 0

vehicle_person['Is_Exceeding_Limit'] = vehicle_person['Speeding_Status_Category'].apply(is_exceeding_limit)
vehicle_person['Is_Too_Fast'] = vehicle_person['Speeding_Status_Category'].apply(is_too_fast_for_conditions)

speeding_counts_agg = vehicle_person.groupby('Crash_ID').agg(
    Total_Exceeding_Limit_Vehicles=('Is_Exceeding_Limit', 'sum'),
    Total_Too_Fast_Vehicles=('Is_Too_Fast', 'sum')
).reset_index()

total_killed_alive = total_killed_alive.merge(
    speeding_counts_agg, 
    on='Crash_ID', 
    how='left'
)
```
